```python
import numpy as np
from transformers import AutoProcessor
fast_tokenizer_path = "physical-intelligence/fast"
fast_tokenizer = AutoProcessor.from_pretrained(fast_tokenizer_path, trust_remote_code=True)
```

```python
from scipy.fft import dct
scale = 10
min_token = 0
action_chunk = np.random.rand(1, 10, 10)
dct_coeff = dct(action_chunk, axis=1, norm="ortho")
dct_coeff = np.around(dct_coeff * scale)
for elem in dct_coeff:
    # ensure all values are positve and convert to chars (works up to the range where coefficients become too large for chr())
    # can handle values which lie significantly outside the [-1, 1] range
    x = list(map(chr, np.maximum(elem.flatten() - min_token, 0).astype(int)))
    print(x)
```

```python
batch_size = 1
timesteps = 10
state_dim = 6
dummy_data = np.random.rand(1, 10, state_dim) * 2 - 1
tokens = fast_tokenizer(action_chunk=dummy_data)
print("num toknes when state is in range:", len(tokens[0]))
tokens = fast_tokenizer(action_chunk=dummy_data * 10)
print("num tokens when state is out of range:", len(tokens[0]))
```
